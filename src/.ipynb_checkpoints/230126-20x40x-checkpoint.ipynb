```python
import os
import random
import numpy as np
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns
import gc 
import time
import copy
import math

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import torchinfo
from torchinfo import summary
from torch.utils.data import Dataset, DataLoader
from torch.utils.data import WeightedRandomSampler
from torchvision import transforms
from torchvision import models
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report, accuracy_score, f1_score, cohen_kappa_score, roc_auc_score, precision_score, recall_score
from PIL import Image

from pytorch_grad_cam import GradCAM
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget
from pytorch_grad_cam.utils.image import show_cam_on_image
```

```python
# ====================== PHẦN 2: CẤU HÌNH & SEED ======================
base_path = '/kaggle/input/lunghist700/LungHist700_combined/data/'
csv_path = base_path + 'data.csv'
images_base = base_path + 'images/' 

batch_size = 32
num_epochs = 30
learning_rate = 0.001
random_seed = 42

def seed_everything(seed):
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)

seed_everything(random_seed)

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Thiết bị đang sử dụng: {device}\n")
```

```python
# ====================== PHẦN 3: LOAD CSV & MAPPING LABEL ======================
full_df = pd.read_csv(csv_path)
superclass_map = {'aca': 0, 'ssc': 1, 'scc': 1, 'nor': 2}
full_df['label'] = full_df['superclass'].map(superclass_map)

print("Đã load và map dữ liệu gốc thành công!")
print("Tổng số dòng:", len(full_df))
print("Các cột:", full_df.columns.tolist())
print("Ví dụ 5 hàng đầu:\n", full_df.head())
print("Phân bố resolution:\n", full_df['resolution'].value_counts())
```

```python
def get_dataframe_no_loss(df_input):
    return df_input.sample(frac=1, random_state=random_seed).reset_index(drop=True)

print("Đã cập nhật hàm xử lý dữ liệu!")
```

```python
# ====================== PHẦN 4: ĐỊNH NGHĨA DATASET CLASS ======================
class LungHistDataset(Dataset):
    def __init__(self, dataframe, root_dir, transform=None):
        self.df = dataframe
        self.root_dir = root_dir
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        superclass = row['superclass'] 
        image_id = row['image_id']
        resolution = row['resolution']
        subclass = row['subclass'] if not pd.isna(row['subclass']) else ""

        folder_name = superclass
        if superclass == 'scc':
            folder_name = 'ssc' 
        
        class_dir = os.path.join(self.root_dir, folder_name)

        potential_filenames = [
            f"{image_id}.jpg",                                      
            f"{superclass}_{subclass}_{resolution}_{image_id}.jpg", 
            f"{superclass}_{resolution}_{image_id}.jpg",            
            f"{image_id}.png",                                     
            f"{superclass}_{subclass}_{resolution}_{image_id}.png"
        ]
        
        img_path = None
        for fname in potential_filenames:
            full_path = os.path.join(class_dir, fname)
            if os.path.exists(full_path):
                img_path = full_path
                break
        
        if img_path is None and os.path.exists(class_dir):
            import glob
            search_pattern = os.path.join(class_dir, f"*{image_id}.*")
            found_files = glob.glob(search_pattern)
            if found_files:
                img_path = found_files[0]

        try:
            if img_path is None:
                raise FileNotFoundError(f"Không tìm thấy ảnh ID {image_id} trong {class_dir}")
            image = Image.open(img_path).convert("RGB")
        except Exception as e:
            image = Image.new('RGB', (224, 224))
            
        label = row['label']
        if self.transform:
            image = self.transform(image)
        return image, label

print("Đã định nghĩa Dataset Class thành công!")
```

```python
# ====================== PHẦN 5.1: TRANSFORMS & CUTMIX=====================
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(224, scale=(0.8, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.5),
    transforms.RandomRotation(30),
    transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

val_test_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])


def cutmix(data, targets, alpha=1.0):
    lam = np.random.beta(alpha, alpha)
    batch_size = data.size(0)
    index = torch.randperm(batch_size, device=data.device)
    shuffled_data = data[index]
    shuffled_targets = targets[index]
    
    bbx1, bby1, bbx2, bby2 = rand_bbox(data.size(), lam)
    data[:, :, bby1:bby2, bbx1:bbx2] = shuffled_data[:, :, bby1:bby2, bbx1:bbx2]
    lam = 1 - ((bbx2 - bbx1) * (bby2 - bby1) / (data.size(-1) * data.size(-2)))
    
    return data, targets, shuffled_targets, lam

def rand_bbox(size, lam):
    W = size[3]
    H = size[2]
    cut_rat = np.sqrt(1. - lam)
    cut_w = int(W * cut_rat)
    cut_h = int(H * cut_rat)

    cx = np.random.randint(W)
    cy = np.random.randint(H)

    bbx1 = np.clip(cx - cut_w // 2, 0, W)
    bby1 = np.clip(cy - cut_h // 2, 0, H)
    bbx2 = np.clip(cx + cut_w // 2, 0, W)
    bby2 = np.clip(cy + cut_h // 2, 0, H)

    return bbx1, bby1, bbx2, bby2
    
print("Đã khai báo Transforms và CutMix.")
```

```python
# ====================== PHẦN 5.2: DEMO AUGMENTATION ======================
rand_idx = random.randint(0, len(full_df) - 1)
demo_ds = LungHistDataset(full_df, root_dir=images_base, transform=None)
original_img, label_idx = demo_ds[rand_idx]

idx_to_class = {0: 'ACA', 1: 'SCC', 2: 'NOR'}
class_name = idx_to_class[label_idx]

rand_idx_2 = random.randint(0, len(full_df) - 1)
img_b, label_b_idx = demo_ds[rand_idx_2]
class_name_b = idx_to_class[label_b_idx]

demo_augmentations = [
    ("RandomResizedCrop", transforms.RandomResizedCrop(224, scale=(0.8, 1.0))),
    ("Horizontal Flip", transforms.Compose([transforms.Resize((224, 224)), transforms.RandomHorizontalFlip(p=1.0)])),
    ("Vertical Flip", transforms.Compose([transforms.Resize((224, 224)), transforms.RandomVerticalFlip(p=1.0)])),
    ("Rotation 30°", transforms.Compose([transforms.Resize((224, 224)), transforms.RandomRotation(30)])),
    ("Color Jitter", transforms.Compose([transforms.Resize((224, 224)), transforms.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.1)]))
]

preprocess = transforms.Compose([transforms.Resize((224, 224)),transforms.ToTensor()])

tensor_a = preprocess(original_img).unsqueeze(0)
tensor_b = preprocess(img_b).unsqueeze(0)

lam = np.random.beta(1.0, 1.0)
bbx1, bby1, bbx2, bby2 = rand_bbox(tensor_a.size(), lam)

tensor_cutmix = tensor_a.clone()
tensor_cutmix[:, :, bbx1:bbx2, bby1:bby2] = tensor_b[:, :, bbx1:bbx2, bby1:bby2]

cutmix_img = transforms.ToPILImage()(tensor_cutmix.squeeze(0))


plt.figure(figsize=(22, 5))

plt.subplot(1, 7, 1)
plt.imshow(original_img)
plt.title(f"ORIGINAL ({class_name})\nSize: {original_img.size}", color='darkblue', fontweight='bold')
plt.axis('off')

for i, (name, transform_func) in enumerate(demo_augmentations):
    aug_img = transform_func(original_img)
    
    plt.subplot(1, 7, i + 2)
    plt.imshow(aug_img)
    plt.title(f"{name}\n(224x224)", fontsize=11)
    plt.axis('off')

plt.subplot(1, 7, 7)
plt.imshow(cutmix_img)
plt.title(f"CutMix\n(Mix with {class_name_b})", fontsize=11, fontweight='bold', color='purple')
plt.axis('off')

img_id = full_df.iloc[rand_idx]['image_id']
plt.suptitle(f"MINH HỌA DATA AUGMENTATION & CUTMIX (Image ID: {img_id})", fontsize=16, y=1.05)
plt.tight_layout()
plt.show()
```

```python
# ====================== PHẦN 6: HÀM THỐNG KÊ  ======================
def print_stats_table(df_name, train_df, val_df, test_df):
    c_train = train_df['label'].value_counts().to_dict()
    c_val   = val_df['label'].value_counts().to_dict()
    c_test  = test_df['label'].value_counts().to_dict()
    idx_to_class = {0: 'ACA', 1: 'SCC', 2: 'NOR'}
    
    print(f"\nDATASET: {df_name}")
    print("-" * 42)
    print(f"{'CLASS':<8} | {'TRAIN':<8} | {'VAL':<8} | {'TEST':<8}")
    print("-" * 42)
    for idx in range(3): 
        name = idx_to_class[idx]
        print(f"{name:<8} | {c_train.get(idx, 0):<8} | {c_val.get(idx, 0):<8} | {c_test.get(idx, 0):<8}")
    print("-" * 42 + "\n")

print("Đã định nghĩa hàm print_stats_table thành công!")
```

```python
# ====================== PHẦN 7: IMBALANCED  ======================
def make_weighted_sampler(df_train, label_col='label'):
    class_counts = df_train[label_col].value_counts().sort_index()
    class_counts = class_counts.values
    class_weights = 1.0 / class_counts 
    sample_weights = [class_weights[label] for label in df_train[label_col]]
    sampler = WeightedRandomSampler(weights=sample_weights,num_samples=len(sample_weights), replacement=True)
    return sampler

print("Đã định nghĩa hàm make_weighted_sampler thành công!")
```

```python
# ====================== PHẦN 8.1: XỬ LÝ DATASET 20X ======================

df_20x = full_df[full_df['resolution'] == '20x'].reset_index(drop=True)
print(f"Dataset 20x gốc: {len(df_20x)} ảnh")

df_20x_full = get_dataframe_no_loss(df_20x)

train_20x, temp_20x = train_test_split(df_20x_full, test_size=0.2, stratify=df_20x_full['label'], random_state=random_seed)
val_20x, test_20x   = train_test_split(temp_20x, test_size=0.5, stratify=temp_20x['label'], random_state=random_seed)

train_20x = train_20x.reset_index(drop=True)
val_20x   = val_20x.reset_index(drop=True)
test_20x  = test_20x.reset_index(drop=True)

train_sampler_20x = make_weighted_sampler(train_20x, label_col='label')

train_ds_20x = LungHistDataset(train_20x, images_base, transform=train_transform)
val_ds_20x   = LungHistDataset(val_20x,   images_base, transform=val_test_transform)
test_ds_20x  = LungHistDataset(test_20x,  images_base, transform=val_test_transform)

train_loader_20x = DataLoader(train_ds_20x, batch_size=batch_size, sampler=train_sampler_20x, shuffle=False, num_workers=2)
val_loader_20x   = DataLoader(val_ds_20x,   batch_size=batch_size, shuffle=False, num_workers=2)
test_loader_20x  = DataLoader(test_ds_20x,  batch_size=batch_size, shuffle=False, num_workers=2)

print_stats_table("20X (FULL DATA - WEIGHTED SAMPLING)", train_20x, val_20x, test_20x)
```

```python
# ====================== PHẦN 8.2: XỬ LÝ DATASET 40X ======================

df_40x = full_df[full_df['resolution'] == '40x'].reset_index(drop=True)
print(f"Dataset 40x gốc: {len(df_40x)} ảnh")

df_40x_full = get_dataframe_no_loss(df_40x)

train_40x, temp_40x = train_test_split(df_40x_full, test_size=0.2, stratify=df_40x_full['label'], random_state=random_seed)
val_40x, test_40x   = train_test_split(temp_40x, test_size=0.5, stratify=temp_40x['label'], random_state=random_seed)

train_40x = train_40x.reset_index(drop=True)
val_40x   = val_40x.reset_index(drop=True)
test_40x  = test_40x.reset_index(drop=True)

train_sampler_40x = make_weighted_sampler(train_40x, label_col='label')

train_ds_40x = LungHistDataset(train_40x, images_base, transform=train_transform)
val_ds_40x   = LungHistDataset(val_40x,   images_base, transform=val_test_transform)
test_ds_40x  = LungHistDataset(test_40x,  images_base, transform=val_test_transform)

train_loader_40x = DataLoader(train_ds_40x, batch_size=batch_size, sampler=train_sampler_40x, shuffle=False, num_workers=2)
val_loader_40x   = DataLoader(val_ds_40x,   batch_size=batch_size, shuffle=False, num_workers=2)
test_loader_40x  = DataLoader(test_ds_40x,  batch_size=batch_size, shuffle=False, num_workers=2)

print_stats_table("40X (FULL DATA - WEIGHTED SAMPLING)", train_40x, val_40x, test_40x)
```

```python
# ====================== PHẦN 9.1: ĐỊNH NGHĨA MODEL ======================
def get_model(model_name, num_classes=3, feature_extract=True):
    model = None
    if model_name == 'resnet50':
        weights = models.ResNet50_Weights.DEFAULT
        model = models.resnet50(weights=weights)
        if feature_extract:
            for param in model.parameters(): param.requires_grad = False
        num_ftrs = model.fc.in_features
        model.fc = nn.Sequential(nn.Dropout(0.5), nn.Linear(num_ftrs, num_classes))

    elif model_name == 'efficientnet_b0':
        weights = models.EfficientNet_B0_Weights.DEFAULT
        model = models.efficientnet_b0(weights=weights)
        if feature_extract:
            for param in model.parameters(): param.requires_grad = False
        num_ftrs = model.classifier[1].in_features 
        model.classifier = nn.Sequential(nn.Dropout(0.3), nn.Linear(num_ftrs, num_classes))

    elif model_name == 'efficientnet_b3':
        weights = models.EfficientNet_B3_Weights.DEFAULT 
        model = models.efficientnet_b3(weights=weights)
        if feature_extract:
            for param in model.parameters(): param.requires_grad = False
        num_ftrs = model.classifier[1].in_features
        model.classifier = nn.Sequential(nn.Dropout(0.4), nn.Linear(num_ftrs, num_classes))
        
    elif model_name == 'densenet121':
        weights = models.DenseNet121_Weights.DEFAULT
        model = models.densenet121(weights=weights)
        if feature_extract:
            for param in model.parameters(): param.requires_grad = False
        num_ftrs = model.classifier.in_features
        model.classifier = nn.Sequential(nn.Dropout(0.5), nn.Linear(num_ftrs, num_classes))
    
    return model.to(device)
```

```python
# ====================== PHẦN 9.2: HÀM TRAINING (ĐÃ TÍCH HỢP CUTMIX) ======================
def train_model(model, dataloaders, criterion, optimizer, scheduler, num_epochs=30, save_name='best_model.pth', model_name='MODEL', cutmix_prob=0.0):
    since = time.time()
    
    val_acc_history = []
    train_loss_history = []
    val_loss_history = []
    
    best_model_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0

    print(f"START TRAINING {model_name} WITH CUTMIX (Prob={cutmix_prob})...")

    for epoch in range(num_epochs):
        print(f'\nEpoch {epoch+1:02d}/{num_epochs}')

        for phase in ['train', 'val']:
            if phase == 'train': model.train()
            else: model.eval()

            running_loss = 0.0
            running_corrects = 0

            for inputs, labels in dataloaders[phase]:
                inputs = inputs.to(device)
                labels = labels.to(device)

                optimizer.zero_grad()

                with torch.set_grad_enabled(phase == 'train'):
                    if phase == 'train' and np.random.rand() < cutmix_prob:
                        inputs, targets_a, targets_b, lam = cutmix(inputs, labels, alpha=1.0)
                        inputs = inputs.to(device)
                        targets_a = targets_a.to(device)
                        targets_b = targets_b.to(device)
                        outputs = model(inputs)
                        loss = criterion(outputs, targets_a) * lam + criterion(outputs, targets_b) * (1. - lam)
                    else:
                        outputs = model(inputs)
                        loss = criterion(outputs, labels)
                    _, preds = torch.max(outputs, 1)

                    if phase == 'train':
                        loss.backward()
                        optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)

            epoch_loss = running_loss / len(dataloaders[phase].dataset)
            epoch_acc = running_corrects.double() / len(dataloaders[phase].dataset)

            phase_label = "Train" if phase == 'train' else "Val  "
            print(f"  {phase_label}  → Loss: {epoch_loss:.4f} | Acc: {epoch_acc:.4f}")

            if phase == 'train':
                train_loss_history.append(epoch_loss)
            else:
                val_loss_history.append(epoch_loss)
                val_acc_history.append(epoch_acc.item())

                if scheduler:
                    if isinstance(scheduler, torch.optim.lr_scheduler.ReduceLROnPlateau): scheduler.step(epoch_loss)
                    else: scheduler.step()

                if epoch_acc > best_acc:
                    best_acc = epoch_acc
                    best_model_wts = copy.deepcopy(model.state_dict())
                    torch.save(model.state_dict(), save_name)
                    print(f">>> BEST MODEL SAVED at epoch {epoch+1} | Val Acc: {best_acc:.4f}")

    time_elapsed = time.time() - since
    print(f'\nTraining complete in {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s')
    print(f'Best Val Acc: {best_acc:.4f}')

    model.load_state_dict(best_model_wts)
    history = {'train_loss': train_loss_history, 'val_loss': val_loss_history, 'val_acc': val_acc_history}
    
    return model, history, best_acc
```

```python
# ====================== PHẦN 9.3: MODEL SUMMARY ======================
model_names_check = ['resnet50', 'efficientnet_b0', 'efficientnet_b3', 'densenet121']
sample_images, sample_labels = next(iter(train_loader_20x)) 
real_input_shape = sample_images.shape 

print(f"Đã lấy được kích thước input thật: {real_input_shape}")

for name in model_names_check:
    print(f"\n   ➤ MODEL: {name.upper()}")

    model_temp = get_model(name, num_classes=3, feature_extract=False)
    summary(model_temp, input_size=real_input_shape, 
            col_names=["input_size", "output_size", "num_params", "mult_adds"],
            depth=1, verbose=1)
    del model_temp
    gc.collect()
    torch.cuda.empty_cache()
```

```python
# ====================== PHẦN 10.1: TRAINING 20X ======================
model_names = ['resnet50', 'efficientnet_b0', 'efficientnet_b3', 'densenet121']

results_20x = {}
histories_20x = {}
trained_models_20x = {}

loaders_20x = {'train': train_loader_20x, 'val': val_loader_20x}
criterion = nn.CrossEntropyLoss()

print(f">>> BẮT ĐẦU SO SÁNH 4 MODEL TRÊN TẬP 20X")
print(f"    Cấu hình: Epochs={num_epochs} | LR={learning_rate} | Batch={batch_size}")
print("="*60)

for name in model_names:
    print(f"\n[{name.upper()}] Đang khởi tạo model...")
    print(f"--- START TRAINING {name.upper()} ---")

    model = get_model(name, num_classes=3, feature_extract=False)
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.1, patience=3)
    
    save_path = f"best_{name}_20x.pth"
    model, hist, best_acc = train_model(model, loaders_20x, criterion, optimizer, scheduler, num_epochs=num_epochs,save_name=save_path,model_name=name.upper(),cutmix_prob=0.0)
    
    results_20x[name] = best_acc
    histories_20x[name] = hist
    trained_models_20x[name] = model 

print("\n" + "="*60)
print("TỔNG KẾT ACCURACY (20X):")
for name, acc in results_20x.items():
    print(f"{name:<15}: {acc:.4f}")
print("="*60)
```

```python
# ====================== PHẦN 10.2: TRAINING 40X ======================
model_names = ['resnet50', 'efficientnet_b0', 'efficientnet_b3', 'densenet121']

results_40x = {}
histories_40x = {}
trained_models_40x = {}

loaders_40x = {'train': train_loader_40x, 'val': val_loader_40x}
criterion = nn.CrossEntropyLoss()

print(f">>> BẮT ĐẦU SO SÁNH 4 MODEL TRÊN TẬP 40X")
print(f"    Cấu hình: Epochs={num_epochs} | LR={learning_rate} | Batch={batch_size}")
print("="*60)

for name in model_names:
    print(f"\n[{name.upper()}] Đang khởi tạo model...")
    print(f"--- START TRAINING {name.upper()} ---")
    
    model = get_model(name, num_classes=3, feature_extract=False)
    optimizer = optim.Adam(model.parameters(), lr=learning_rate)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.1, patience=3)

    save_path = f"best_{name}_40x.pth"
    model, hist, best_acc = train_model(model, loaders_40x, criterion, optimizer, scheduler, num_epochs=num_epochs,save_name=save_path,model_name=name.upper(),cutmix_prob=0.0)
    
    results_40x[name] = best_acc
    histories_40x[name] = hist
    trained_models_40x[name] = model

print("\n" + "="*60)
print("TỔNG KẾT ACCURACY (40X):")
for name, acc in results_40x.items():
    print(f"{name:<15}: {acc:.4f}")
print("="*60)
```

```python
# ====================== PHẦN 11: PLOT ======================
def plot_line_chart_comparison(histories, title_suffix=""):
    sns.set_style("whitegrid") 
    colors = ['#1f77b4', '#d62728', '#2ca02c', '#ff7f0e', '#9467bd']
    
    fig, axes = plt.subplots(1, 2, figsize=(15, 5))
    
    ax1 = axes[0]
    for i, (name, hist) in enumerate(histories.items()):
        val_acc = hist['val_acc']
        epochs = range(1, len(val_acc) + 1)
        ax1.plot(epochs, val_acc, label=name, color=colors[i % len(colors)], linewidth=1.5, linestyle='-', alpha=0.9)       
        best_val = max(val_acc)
        best_epoch = val_acc.index(best_val) + 1
        ax1.scatter(best_epoch, best_val, s=150, c=colors[i], edgecolors='k', zorder=10)
    ax1.set_title(f'VALIDATION ACCURACY {title_suffix}', fontsize=14, fontweight='bold', color='#333333')
    ax1.set_xlabel('Epochs', fontsize=12)
    ax1.set_ylabel('Accuracy', fontsize=12)
    ax1.legend(fontsize=11, frameon=True, fancybox=True, shadow=True)

    ax2 = axes[1]
    for i, (name, hist) in enumerate(histories.items()):
        val_loss = hist['val_loss']
        epochs = range(1, len(val_loss) + 1)
        ax2.plot(epochs, val_loss, label=name, color=colors[i % len(colors)],linewidth=1.5, linestyle='-',alpha=0.9)
        min_loss = min(val_loss)
        best_epoch_loss = val_loss.index(min_loss) + 1
        ax2.scatter(best_epoch_loss, min_loss, s=150, c=colors[i], edgecolors='k', zorder=10)
    ax2.set_title(f'VALIDATION LOSS {title_suffix}', fontsize=14, fontweight='bold', color='#333333')
    ax2.set_xlabel('Epochs', fontsize=12)
    ax2.set_ylabel('Loss', fontsize=12)
    ax2.legend(fontsize=11, frameon=True, fancybox=True, shadow=True)

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.3)
    plt.show()

print(f"BIỂU ĐỒ SO SÁNH TẬP 20X")
plot_line_chart_comparison(histories_20x, title_suffix="(Dataset 20X)")
print(f"\n\nBIỂU ĐỒ SO SÁNH TẬP 40X")
plot_line_chart_comparison(histories_40x, title_suffix="(Dataset 40X)")
```

```python
# ====================== PHẦN 12: EVALUATION ======================
def evaluate_model_full(model, loader, model_name, class_names=['ACA', 'SSC', 'NOR']):
    model.eval()
    y_true = []
    y_pred = []
    y_probs = []
    
    print(f"\nREPORT: {model_name.upper()}")
    
    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(inputs)
            outputs = torch.nan_to_num(outputs, nan=0.0, posinf=10.0, neginf=-10.0)
            _, preds = torch.max(outputs, 1)
            probs = F.softmax(outputs, dim=1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())
            y_probs.extend(probs.cpu().numpy())

    y_probs = np.array(y_probs)
    y_probs = np.nan_to_num(y_probs, nan=1.0/len(class_names))

    acc = accuracy_score(y_true, y_pred)
    kappa = cohen_kappa_score(y_true, y_pred)

    try:
        labels_index = list(range(len(class_names))) 
        roc_auc = roc_auc_score(y_true, y_probs, multi_class='ovr', average='macro', labels=labels_index)
    except Exception as e:
        print(f"LỖI ROC-AUC: {e}")
        print(f"(Shape y_probs: {y_probs.shape} | Unique y_true: {np.unique(y_true)})")
        roc_auc = 0.0

    print(f"   ➤ TEST RESULT: Accuracy: {acc:.4f} | ROC-AUC (Macro): {roc_auc:.4f} | Cohen's Kappa: {kappa:.4f} \n")
    
    print("CLASSIFICATION REPORT:")
    print(classification_report(y_true, y_pred, target_names=class_names, digits=4, zero_division=0))
    print("-" * 85)
    
    return acc, roc_auc, kappa
```

```python
# ====================== PHẦN 12.1: TEST RESULT 20X ======================
print("KẾT QUẢ ĐÁNH GIÁ (DATASET 20X)")
for name, model in trained_models_20x.items(): evaluate_model_full(model, test_loader_20x, f"{name}")
```

```python
# ====================== PHẦN 13: STATISTICS OF RESULTS ======================
def get_metrics_for_table(model_name, loader, weight_path, num_classes=3):
    model = get_model(model_name, num_classes=num_classes, feature_extract=False)
    try:
        model.load_state_dict(torch.load(weight_path, map_location=device))
    except FileNotFoundError:
        print(f"Không tìm thấy file weight: {weight_path}")
        return None 
        
    model.to(device)
    model.eval()
    
    y_true = []
    y_pred = []
    y_probs = []

    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(inputs)
            outputs = torch.nan_to_num(outputs, nan=0.0, posinf=10.0, neginf=-10.0)
            probs = F.softmax(outputs, dim=1)
            _, preds = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())
            y_probs.extend(probs.cpu().numpy())

    y_probs = np.array(y_probs)
    y_probs = np.nan_to_num(y_probs, nan=1.0/num_classes)

    acc = accuracy_score(y_true, y_pred)
    
    try:
        auc = roc_auc_score(y_true, y_probs, multi_class='ovr', average='weighted', labels=list(range(num_classes)))
    except Exception as e:
        print(f"LỖI ROC-AUC {model_name}: {e}")
        auc = 0.0
        
    prec = precision_score(y_true, y_pred, average='macro', zero_division=0)
    rec = recall_score(y_true, y_pred, average='macro', zero_division=0)
    
    return acc, auc, prec, rec

ref_data = [
    ["DNN (ResNet50)", "20x", 90.00, 98.00, 92.00, 87.00],
    ["DNN (ResNet50)", "40x", 82.00, 94.00, 82.00, 84.00],
    ["MIL", "40x", 81.00, 89.00, 80.00, 81.00],
]
df_ref = pd.DataFrame(ref_data, columns=["Model", "Resolution", "Accuracy (%)", "ROC-AUC (%)", "Precision (%)", "Recall (%)"])

ours_data = []
my_models = ['resnet50', 'efficientnet_b0', 'efficientnet_b3', 'densenet121']

for name in my_models:
    path = f"best_{name}_20x.pth" 
    metrics = get_metrics_for_table(name, test_loader_20x, path) 
    if metrics:
        acc, auc, prec, rec = metrics
        display_name = name.replace('_', ' ').title() 
        ours_data.append([display_name, "20x", acc*100, auc*100, prec*100, rec*100])

if 'test_loader_40x' in globals():
    for name in my_models:
        path = f"best_{name}_40x.pth"
        metrics = get_metrics_for_table(name, test_loader_40x, path)
        if metrics:
            acc, auc, prec, rec = metrics
            display_name = name.replace('_', ' ').title()
            ours_data.append([display_name, "40x", acc*100, auc*100, prec*100, rec*100])

df_ours = pd.DataFrame(ours_data, columns=["Model", "Resolution", "Accuracy (%)", "ROC-AUC (%)", "Precision (%)", "Recall (%)"])
df_ours = df_ours.sort_values(by=['Resolution', 'Accuracy (%)'], ascending=[True, False])


print("\n" + "-"*22 + " BẢNG 1: KẾT QUẢ THAM KHẢO " + "-"*22)
display(df_ref.round(2))
print("\n\n" + "-"*20 + " BẢNG 2: KẾT QUẢ THỰC NGHIỆM " + "-"*20)
display(df_ours.round(2))
```

```python
# ====================== PHẦN 14: SELECT & LOAD BEST MODELS ======================
model_list = ['resnet50', 'efficientnet_b0', 'efficientnet_b3', 'densenet121']

def find_best_model_auto(loader, resolution, model_names):
    candidates = {}
    for name in model_names:
        weight_path = f"best_{name}_{resolution}.pth"
        if not os.path.exists(weight_path):
            continue
        model = get_model(name, num_classes=3, feature_extract=False)
        try:
            model.load_state_dict(torch.load(weight_path, map_location=device))
        except:
            print(f"Lỗi load weight {name}")
            continue
            
        model.to(device)
        model.eval()

        y_true = []
        y_pred = []
        y_probs = []
        
        with torch.no_grad():
            for inputs, labels in loader:
                inputs = inputs.to(device)
                outputs = model(inputs)
                outputs = torch.nan_to_num(outputs, nan=0.0)
                probs = F.softmax(outputs, dim=1)
                _, preds = torch.max(outputs, 1)
                y_true.extend(labels.cpu().numpy())
                y_pred.extend(preds.cpu().numpy())
                y_probs.extend(probs.cpu().numpy())

        acc = accuracy_score(y_true, y_pred)
        kappa = cohen_kappa_score(y_true, y_pred)
        roc_auc = roc_auc_score(y_true, np.array(y_probs), multi_class='ovr', average='macro')
        
        candidates[name] = {'acc': acc, 'kappa': kappa, 'roc-auc': roc_auc}

    if not candidates:
        return None, None, None

    best_name = max(candidates, key=lambda x: (candidates[x]['kappa'], candidates[x]['acc'], candidates[x]['roc-auc']))
    best_stats = candidates[best_name]
    
    return best_name, best_stats, candidates


best_name_20x, best_stats_20x, _ = find_best_model_auto(test_loader_20x, "20x", model_list)
print(f"1. DATASET 20X:")
if best_name_20x:
    print(f"Best Model: {best_name_20x.upper()}")
    print(f"Kappa:      {best_stats_20x['kappa']:.4f}")
    print(f"Accuracy:   {best_stats_20x['acc']:.4f}")
    print(f"ROC-AUC:    {best_stats_20x['roc-auc']:.4f}")
    final_model_20x = get_model(best_name_20x, num_classes=3, feature_extract=False)
    final_model_20x.load_state_dict(torch.load(f"best_{best_name_20x}_20x.pth", map_location=device))
    final_model_20x.to(device)
    final_model_20x.eval()
    print("   ➤ Đã nạp weights thành công vào biến 'final_model_20x'")


best_name_40x, best_stats_40x, _ = find_best_model_auto(test_loader_40x, "40x", model_list)
print("-"*60)
print(f"2. DATASET 40X:")
if best_name_40x:
    print(f"Best Model: {best_name_40x.upper()}")
    print(f"Kappa:      {best_stats_40x['kappa']:.4f}")
    print(f"Accuracy:   {best_stats_40x['acc']:.4f}")
    print(f"ROC-AUC:    {best_stats_40x['roc-auc']:.4f}")
    final_model_40x = get_model(best_name_40x, num_classes=3, feature_extract=False)
    final_model_40x.load_state_dict(torch.load(f"best_{best_name_40x}_40x.pth", map_location=device))
    final_model_40x.to(device)
    final_model_40x.eval()
    print("   ➤ Đã nạp weights thành công vào biến 'final_model_40x'")
```

```python
# ====================== PHẦN 15: CONFUSION MATRIX ======================
def get_prediction_data(model, loader):
    model.eval()
    y_true = []
    y_pred = []
    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())  
    return y_true, y_pred

def plot_dual_confusion_matrix(model20, loader20, name20, model40, loader40, name40):
    class_names = ['ACA', 'NOR', 'SSC'] 
    
    y_true_20, y_pred_20 = get_prediction_data(model20, loader20)
    cm_20 = confusion_matrix(y_true_20, y_pred_20)
    acc_20 = accuracy_score(y_true_20, y_pred_20)
    kap_20 = cohen_kappa_score(y_true_20, y_pred_20)

    y_true_40, y_pred_40 = get_prediction_data(model40, loader40)
    cm_40 = confusion_matrix(y_true_40, y_pred_40)
    acc_40 = accuracy_score(y_true_40, y_pred_40)
    kap_40 = cohen_kappa_score(y_true_40, y_pred_40)
    

    fig, axes = plt.subplots(1, 2, figsize=(14, 7))

    sns.heatmap(cm_20, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[0], annot_kws={"size": 16, "weight": "bold"}, cbar=False)
    axes[0].set_xlabel('\nPredicted Label', fontsize=12)
    axes[0].set_ylabel('True Label\n', fontsize=12)
    title_20 = (f"Model: {name20.upper()} (20x)\n"
                f"Acc: {acc_20:.2%} | Kappa: {kap_20:.4f}")
    axes[0].set_title(title_20, fontsize=14, fontweight='bold', color='#1f77b4', pad=15)

    sns.heatmap(cm_40, annot=True, fmt='d', cmap='Greens', xticklabels=class_names, yticklabels=class_names, ax=axes[1], annot_kws={"size": 16, "weight": "bold"}, cbar=False)
    axes[1].set_xlabel('\nPredicted Label', fontsize=12)
    axes[1].set_ylabel('True Label\n', fontsize=12)
    title_40 = (f"Model: {name40.upper()} (40x)\n"
                f"Acc: {acc_40:.2%} | Kappa: {kap_40:.4f}")
    axes[1].set_title(title_40, fontsize=14, fontweight='bold', color='#2ca02c', pad=15)

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.3)
    plt.show()

plot_dual_confusion_matrix(final_model_20x, test_loader_20x, best_name_20x, final_model_40x, test_loader_40x, best_name_40x)
```

```python
# ====================== PHẦN 16: GRAD-CAM ======================
IMG_SIZE = 2.8

def get_target_layer(model, model_name):
    name_lower = model_name.lower()
    if 'resnet' in name_lower: return [model.layer4[-1]]
    elif 'efficientnet' in name_lower: return [model.features[-1]]
    elif 'densenet' in name_lower: return [model.features[-1]]
    else: return [model.features[-1]]

def denormalize_image(tensor):
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    img = tensor.permute(1, 2, 0).cpu().numpy()
    img = std * img + mean
    img = np.clip(img, 0, 1)
    return img

def visualize_3_classes_correct(model, model_name, loader):
    model.eval()
    target_layers = get_target_layer(model, model_name)
    cam = GradCAM(model=model, target_layers=target_layers)
    
    class_names = ['ACA', 'NOR', 'SSC']
    found_images = {} 
    needed_classes = {0, 1, 2}
    
    print(f"➤ Hiển thị Grad-CAM cho ảnh dự đoán đúng ({model_name.upper()})\n")
    
    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            
            for i in range(len(labels)):
                lbl = labels[i].item()
                pred = preds[i].item()
                if lbl == pred: 
                    if lbl in needed_classes and lbl not in found_images:
                        found_images[lbl] = (inputs[i].cpu(), lbl, pred)
            if len(found_images) == 3:
                break
    
    if len(found_images) < 3:
        print(f"Cảnh báo: Chỉ tìm thấy đại diện đúng cho các lớp: {list(found_images.keys())}")
        ordered_selection = list(found_images.values())
    else:
        ordered_selection = [found_images[i] for i in range(3)]

    orig_imgs_plot = []
    heatmap_imgs_plot = []
    titles_orig = []
    titles_heat = []

    for img_tensor, label_idx, pred_idx in ordered_selection:
        input_tensor = img_tensor.unsqueeze(0).to(device)
        targets = [ClassifierOutputTarget(pred_idx)]
        grayscale_cam = cam(input_tensor=input_tensor, targets=targets)[0, :]
        rgb_img = denormalize_image(img_tensor)
        visualization = show_cam_on_image(rgb_img, grayscale_cam, use_rgb=True)
        orig_imgs_plot.append(rgb_img)
        heatmap_imgs_plot.append(visualization)
        label_str = class_names[label_idx]
        titles_orig.append(f"True: {label_str}")
        titles_heat.append(f"Pred: {label_str} (Correct)")
    
    num_cols = len(ordered_selection)
    fig, axes = plt.subplots(2, num_cols, figsize=(num_cols * IMG_SIZE, 2 * IMG_SIZE))
    
    if num_cols == 1:
        axes = axes.reshape(2, 1)
    for i in range(num_cols):
        axes[0, i].imshow(orig_imgs_plot[i])
        axes[0, i].set_title(titles_orig[i], fontsize=14, fontweight='bold', color='black')
        axes[0, i].axis('off')
    for i in range(num_cols):
        axes[1, i].imshow(heatmap_imgs_plot[i])
        axes[1, i].set_title(titles_heat[i], fontsize=14, fontweight='bold', color='green')
        axes[1, i].axis('off')
        class_id = ordered_selection[i][1]
        axes[1, i].set_xlabel(f"Class: {class_names[class_id]}")

    plt.tight_layout()
    plt.subplots_adjust(hspace=0.2)
    plt.show()

print("\nHEATMAP REPORT (20X)")
visualize_3_classes_correct(final_model_20x, best_name_20x, test_loader_20x)
print("-" * 150)
print("\nHEATMAP REPORT (40X)")
visualize_3_classes_correct(final_model_40x, best_name_40x, test_loader_40x)
```

```python
# ====================== PHẦN 17: ERROR VISUALIZATION ======================
def visualize_errors_top_bottom(model, model_name, loader):
    model.eval()
    target_layers = get_target_layer(model, model_name)
    cam = GradCAM(model=model, target_layers=target_layers)
    
    class_names = ['ACA', 'SSC', 'NOR']
    pool_wrong = []
    
    with torch.no_grad():
        for inputs, labels in loader:
            inputs = inputs.to(device)
            labels = labels.to(device)
            outputs = model(inputs)
            _, preds = torch.max(outputs, 1)
            mismatches = (preds != labels).nonzero(as_tuple=True)[0]
            for idx in mismatches:
                pool_wrong.append((inputs[idx].cpu(), labels[idx].cpu(), preds[idx].cpu()))
    total_errors = len(pool_wrong)
    if total_errors == 0:
        print(f" Model {model_name} không có mẫu nào sai.")
        return
    print(f"   ➤ Đã tìm thấy {total_errors} mẫu sai của {model_name.upper()}.\n")
    

    images_per_row = 6
    num_blocks = math.ceil(total_errors / images_per_row)
    total_plt_rows = num_blocks * 2
    fig_height = 3 * total_plt_rows 
    fig_width = 16
    
    fig, axes = plt.subplots(total_plt_rows, images_per_row, figsize=(fig_width, fig_height))
    if total_plt_rows == 1: axes = axes.reshape(1, -1)
    if images_per_row == 1: axes = axes.reshape(-1, 1)
    if len(axes.shape) == 1: axes = axes.reshape(-1, images_per_row)
    for ax in axes.flatten():
        ax.axis('off')

    orig_imgs_plot = []
    heatmap_imgs_plot = []
    titles_orig = []
    titles_heat = []

    for i, (img_tensor, label_idx, pred_idx) in enumerate(pool_wrong):
        block_idx = i // images_per_row    
        col_idx = i % images_per_row      
        row_orig_idx = block_idx * 2        
        row_heat_idx = block_idx * 2 + 1     

        input_tensor = img_tensor.unsqueeze(0).to(device)
        targets = [ClassifierOutputTarget(pred_idx.item())]
        grayscale_cam = cam(input_tensor=input_tensor, targets=targets)[0, :]
        rgb_img = denormalize_image(img_tensor)
        visualization = show_cam_on_image(rgb_img, grayscale_cam, use_rgb=True)
        label_str = class_names[label_idx.item()]
        pred_str = class_names[pred_idx.item()]

        ax_orig = axes[row_orig_idx, col_idx]
        ax_orig.imshow(rgb_img)
        ax_orig.set_title(f"True: {label_str}", fontsize=14, fontweight='bold', color='black')
        ax_orig.axis('off')
        
        ax_heat = axes[row_heat_idx, col_idx]
        ax_heat.imshow(visualization)
        ax_heat.set_title(f"Pred: {pred_str}", fontsize=14, fontweight='bold', color='red') 
        ax_heat.axis('off')

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.1)
    plt.show()

print("\nERROR ANALYSIS REPORT: DATASET 20X")
visualize_errors_top_bottom(final_model_20x, best_name_20x, test_loader_20x)
print("\n" + "-"*150)
print("\nERROR ANALYSIS REPORT: DATASET 40X")
visualize_errors_top_bottom(final_model_40x, best_name_40x, test_loader_40x)
```
