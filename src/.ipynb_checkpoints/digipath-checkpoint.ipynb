```python

def visualize_histogram(pil_img, title="Histogram"):
    img = np.array(pil_img)
    lab = cv2.cvtColor(img, cv2.COLOR_RGB2LAB)
    L = lab[:, :, 0]  # Extract L channel
    
    plt.figure(figsize=(10, 5))
    plt.hist(L.flatten(), bins=256, color='gray', alpha=0.7)
    plt.axvline(L.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: {L.mean():.1f}')
    plt.title(title, fontsize=14)
    plt.xlabel("Brightness (0-255)")
    plt.ylabel("Number of Pixels")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.xlim([0, 256])
    plt.tight_layout()
    # plt.savefig(f'{title}.png')
    plt.show()

visualize_histogram(pil_img, "Original Image Histogram")
visualize_histogram(clahe_img, "CLAHE Enhanced Histogram")
```

```python
det_transform_pil = transforms.Compose(
    [
        transforms.Resize(size=(224, 224), antialias=True),
        transforms.Lambda(clahe_transform),
    ]
)

rand_transform_pil = transforms.Compose(
    [
        transforms.Resize(size=(224, 224), antialias=True),
        transforms.Lambda(clahe_transform),
        transforms.RandomHorizontalFlip(p=0.3),
        transforms.RandomVerticalFlip(p=0.5),
        transforms.RandomApply([transforms.ColorJitter(0.4, 0.4, 0.4, 0.1)], p=0.8),
        transforms.RandomGrayscale(p=0.2),
    ]
)
display(pil_img.resize((224, 224)))
display(rand_transform_pil(pil_img))
display(det_transform_pil(pil_img))
```

### Grad-CAM Visualization

```python
from grad_cam import run_gradcam
import os
import unittest
from pathlib import Path

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torchvision
from PIL import Image
from pytorch_grad_cam import GradCAM
from pytorch_grad_cam.utils.image import show_cam_on_image
from pytorch_grad_cam.utils.model_targets import ClassifierOutputTarget
from torchvision.models import ViT_B_16_Weights

from constant import (
    GRADCAM_OUT_DIR,
    LUNG_IMAGES_DIR,
    LUNG_LOADED_FILE,
    LUNG_METADATA_FILE,
    LUNG_MODEL_DIR,
)
from dataset import load_dataset
from model import ClassifierHead
from transform import det_transform
```

```python
csv_file=LUNG_METADATA_FILE
root_dir=LUNG_IMAGES_DIR
load_file=LUNG_LOADED_FILE

df = pd.read_csv(load_file, encoding="utf-8")
_, _, _, label_map = load_dataset(
    csv_file, root_dir, load_file
)

idx_label_map = {v: k for k, v in label_map.items()}
```

```python
# Must run a single training beforehand to have the model checkpoint at LUNG_MODEL_DIR/LUNG_ClassifierHead_199_EPOCHS.pth.tar
def demo_test_split():
    train_idx, valid_idx, test_idx, label_map = load_dataset(
        LUNG_METADATA_FILE, LUNG_IMAGES_DIR, LUNG_LOADED_FILE
    )
    df = pd.read_csv(LUNG_LOADED_FILE, encoding="utf-8")
    device = "cuda" if torch.cuda.is_available() else "cpu"

    ckpt = Path(LUNG_MODEL_DIR, "LUNG_ClassifierHead_199_EPOCHS.pth.tar")
    if os.path.exists(ckpt):
        latest_pil, latest_grad_pil, label = run_gradcam(
            df,
            test_idx,
            label_map,
            ckpt,
            n_samples=1,
            device=device,
            stain_normalizer=None,
        )
    else:
        print(f"=> ERROR: Model head checkpoint {ckpt} not found") 
    return latest_pil, latest_grad_pil, label

pil, grad, label = demo_test_split()
```

```python
print("\n=> Control Stain Sample Image\n")
display(pil.resize((224, 224)))
print("\n=> Grad-CAM Stain Sample Image\n")
display(grad)
print(f"\n=> Sample Grad-CAM Visualization of label \"{idx_label_map[label]}\"")
```
